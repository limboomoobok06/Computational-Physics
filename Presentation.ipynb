```python

import math

# Physical constants
h = 6.62607015e-34      # Planck's constant (J s)
c = 2.99792458e8        # Speed of light (m/s)
kB = 1.380649e-23       # Boltzmann constant (J/K)

# Step 1: Define the equation
# f(x) = exp(x)(x - 5) + 5 = 0

def f(x):
    return math.exp(x) * (x - 5) + 5

# Derivative of f(x)
def df(x):
    return math.exp(x) * (x - 4)

# Step 2: Newton-Raphson method
x = 5.0                 # Initial guess
tolerance = 1e-10
max_iterations = 100

for i in range(max_iterations):
    x_new = x - f(x) / df(x)

    if abs(x_new - x) < tolerance:
        x = x_new
        break

    x = x_new

print("Root x =", x)

# Step 3: Calculate Wien's displacement constant
b = h * c / (x * kB)

print("Wien's displacement constant b =", b, "m K")

# Step 4: Find peak wavelength at a given temperature
T = float(input("Enter temperature in Kelvin: "))

if T > 0:
    lambda_peak = b / T

    print("Temperature =", T, "K")
    print("Peak wavelength =", lambda_peak, "m")
    print("Peak wavelength =", lambda_peak * 1e9, "nm")
else:
    print("Temperature must be greater than zero.")
```
